```python

# ============================================================
# CUSTOMER CHURN PREDICTION - EXPLORATORY DATA ANALYSIS
# Python 3.10
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

DATA_FILE = "WA_Fn-UseC_-Telco-Customer-Churn.csv"

# Support both project-root and notebook-folder working directories
possible_paths = [
    Path("data/raw") / DATA_FILE,
    Path("../data/raw") / DATA_FILE
]

DATA_PATH = next(
    (path for path in possible_paths if path.exists()),
    None
)

if DATA_PATH is None:
    raise FileNotFoundError(
        "Dataset not found. Check data/raw/ and the CSV filename."
    )

OUTPUT_DIR = Path("reports/figures")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2. Load dataset
# ------------------------------------------------------------

df = pd.read_csv(DATA_PATH)

print("=" * 55)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 55)

print("Dataset shape:", df.shape)

print("\nFirst five rows:")
display(df.head())

# ------------------------------------------------------------
# 3. Dataset information
# ------------------------------------------------------------

print("\nDATASET INFORMATION")
df.info()

print("\nSTATISTICAL SUMMARY")
display(df.describe(include="all"))

# ------------------------------------------------------------
# 4. Convert TotalCharges into numeric
# ------------------------------------------------------------

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

# ------------------------------------------------------------
# 5. Numerical and categorical columns
# ------------------------------------------------------------

numerical_columns = df.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_columns = df.select_dtypes(
    exclude=["number"]
).columns.tolist()

print("\nNUMERICAL COLUMNS:")
print(numerical_columns)

print("\nCATEGORICAL COLUMNS:")
print(categorical_columns)

# ------------------------------------------------------------
# 6. Missing value analysis
# ------------------------------------------------------------

missing_values = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": df.isnull().mean() * 100
})

print("\nMISSING VALUE ANALYSIS")

display(
    missing_values[
        missing_values["Missing Count"] > 0
    ]
)

# ------------------------------------------------------------
# 7. Duplicate analysis
# ------------------------------------------------------------

print("\nDUPLICATE RECORDS:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 8. Target variable analysis
# ------------------------------------------------------------

print("\nCHURN DISTRIBUTION")
display(df["Churn"].value_counts())

print("\nCHURN PERCENTAGE")
display(df["Churn"].value_counts(normalize=True) * 100)

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Churn")

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.savefig(OUTPUT_DIR / "churn_distribution.png")
plt.show()

# ------------------------------------------------------------
# 9. Churn percentage chart
# ------------------------------------------------------------

churn_percentage = df["Churn"].value_counts(normalize=True) * 100

plt.figure(figsize=(7, 5))

plt.pie(
    churn_percentage,
    labels=churn_percentage.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Churn Percentage")

plt.tight_layout()
plt.savefig(OUTPUT_DIR / "churn_percentage.png")
plt.show()

# ------------------------------------------------------------
# 10. Numerical feature distributions
# ------------------------------------------------------------

numeric_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

df[numeric_features].hist(
    figsize=(12, 7),
    bins=30
)

plt.suptitle("Numerical Feature Distributions")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "numerical_distributions.png")
plt.show()

# ------------------------------------------------------------
# 11. Monthly charges vs churn
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="MonthlyCharges"
)

plt.title("Monthly Charges vs Churn")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "monthly_charges.png")
plt.show()

# ------------------------------------------------------------
# 12. Tenure vs churn
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="tenure"
)

plt.title("Customer Tenure vs Churn")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "tenure_churn.png")
plt.show()

# ------------------------------------------------------------
# 13. Categorical feature analysis
# ------------------------------------------------------------

categorical_features = [
    "Contract",
    "InternetService",
    "PaymentMethod"
]

for feature in categorical_features:

    plt.figure(figsize=(9, 5))

    sns.countplot(
        data=df,
        x=feature,
        hue="Churn"
    )

    plt.title("Churn by " + feature)
    plt.xticks(rotation=20)

    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR / (feature.lower() + "_churn.png")
    )

    plt.show()

# ------------------------------------------------------------
# 14. Correlation analysis
# ------------------------------------------------------------

correlation_data = df[
    [
        "SeniorCitizen",
        "tenure",
        "MonthlyCharges",
        "TotalCharges"
    ]
].copy()

correlation_data["Churn"] = (
    df["Churn"].map({"No": 0, "Yes": 1})
)

correlation_matrix = correlation_data.corr()

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Feature Correlation Heatmap")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "correlation.png")
plt.show()

# ------------------------------------------------------------
# 15. Churn rate by important features
# ------------------------------------------------------------

features_to_analyse = [
    "Contract",
    "InternetService",
    "PaymentMethod",
    "OnlineSecurity",
    "TechSupport"
]

for feature in features_to_analyse:

    print("\nCHURN RATE BY", feature)

    churn_rate = (
        df.groupby(feature)["Churn"]
        .apply(lambda x: (x == "Yes").mean() * 100)
        .sort_values(ascending=False)
    )

    display(churn_rate.to_frame("Churn Rate (%)"))

# ------------------------------------------------------------
# 16. Final EDA summary
# ------------------------------------------------------------

total_customers = len(df)

churned_customers = (df["Churn"] == "Yes").sum()

retained_customers = (df["Churn"] == "No").sum()

churn_percentage = (
    churned_customers / total_customers
) * 100

print("\n" + "=" * 55)
print("CUSTOMER CHURN - EDA SUMMARY")
print("=" * 55)

print("Total customers:", total_customers)
print("Churned customers:", churned_customers)
print("Retained customers:", retained_customers)

print("Churn percentage:", round(churn_percentage, 2), "%")

print("Duplicate records:", df.duplicated().sum())

print("\nEDA COMPLETED SUCCESSFULLY!")
print("Charts saved in:", OUTPUT_DIR.resolve())
```
